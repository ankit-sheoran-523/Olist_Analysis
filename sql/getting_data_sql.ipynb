## orders Table

```python
orders=pd.read_csv("D:\📝ImarticusLearning\project\sql+python\dataset\olist_orders_dataset.csv")
orders = orders.where(pd.notnull(orders), None)
orders.shape
```

```python
orders.dtypes
```

```python
order_t_query="""
create table orders(
order_id varchar(50) primary key,
cust_id varchar(50),
order_status varchar(20),
order_purchase_timestamp datetime,
order_approved_at datetime,
order_delivered_carrier_date datetime,
order_delivered_cust_date datetime,
order_est_delivery_date datetime
);
"""

cur.execute(order_t_query)
db.commit()
```

```python
for _, row in orders.iterrows():
    cur.execute("""
        INSERT IGNORE INTO orders 
        (order_id, cust_id, order_status, order_purchase_timestamp,
         order_approved_at, order_delivered_carrier_date,
         order_delivered_cust_date, order_est_delivery_date)
        VALUES (%s,%s,%s,%s,%s,%s,%s,%s)
    """, tuple(row))

db.commit()
```

## order items

```python
cur.execute("""
create table order_items(
order_item_id int,
order_id varchar(50),
prod_id varchar(50),
seller_id varchar(50),
shipping_limit_date datetime,
price decimal(10,2),
freight_value decimal(10,2),
primary key (order_id , order_item_id));""")
db.commit()
```

```python
order_items=pd.read_csv("dataset\olist_order_items_dataset.csv")
order_items = order_items.where(pd.notnull(order_items), None)
order_items.rename(columns={'product_id': 'prod_id'}, inplace=True)
order_items.shape
```

```python
for _, row in order_items.iterrows():
    cur.execute("""
        INSERT IGNORE INTO order_items
        (order_item_id, order_id, prod_id, seller_id,
         shipping_limit_date, price, freight_value)
        VALUES (%s,%s,%s,%s,%s,%s,%s)
    """, (
        row['order_item_id'],
        row['order_id'],
        row['prod_id'],
        row['seller_id'],
        row['shipping_limit_date'],
        row['price'],
        row['freight_value']
    ))

db.commit()
```
