```python
sns.set_style('darkgrid')
plt.figure(figsize=(25,6))

plt.subplot(1,2,1)
x1=df['seller_city'].value_counts().reset_index().sort_values(by='count',ascending=False)
x1=x1.iloc[:10,:]
ax=sns.barplot(data=x1,x='seller_city',y='count',color='purple')
plt.bar_label(ax.containers[0])
plt.title("Top 10 Cities by Sellers")

plt.subplot(1,2,2)
x2=df['seller_state'].value_counts().reset_index().sort_values(by='count',ascending=False)
ax=sns.barplot(data=x2,x='seller_state',y='count',color='purple')
plt.bar_label(ax.containers[0])
plt.title("States by Sellers count")

plt.tight_layout(pad=2)
plt.savefig('sellers_insights.png',dpi=500,bbox_inches="tight")
plt.show()
```
